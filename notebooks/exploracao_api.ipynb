```python
import json
from datetime import datetime, timezone

import requests

url = (
    "https://dadosabertos.compras.gov.br"
    "/modulo-pesquisa-preco/1_consultarMaterial"
)

parametros = {
    "tipo": "codigoItemCatalogo",
    "codigo": 217129,
    "pagina": 1,
    "tamanhoPagina": 100,
}

horario_consulta = datetime.now(timezone.utc).isoformat()

# mostra o endereço completo mesmo se houver falha de conexão
url_completa = requests.Request(
    "GET", url, params=parametros
).prepare().url

print("Horário UTC:", horario_consulta)
print("Código consultado:", parametros["codigo"])
print("URL:", url_completa)

try:
    resposta = requests.get(
        url,
        params=parametros,
        timeout=60,
    )

    print("Status HTTP:", resposta.status_code)

    if resposta.status_code == 200:
        dados = resposta.json()
        registros = dados.get("resultado", [])

        print("Registros nesta página:", len(registros))
        print("Total informado pela API:", dados.get("totalRegistros"))

        if registros:
            print("\nPrimeiro registro:")
            print(json.dumps(registros[0], indent=2, ensure_ascii=False))
        else:
            print("A API respondeu, mas não retornou registros.")

    else:
        print("A consulta não retornou HTTP 200.")
        print("Trecho da resposta:", resposta.text[:500])

except requests.exceptions.RequestException as erro:
    print("Falha na requisição:", type(erro).__name__, str(erro))

except ValueError as erro:
    print("Não foi possível interpretar a resposta como JSON:", str(erro))
```
